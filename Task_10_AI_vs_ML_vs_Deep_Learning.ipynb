```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 6))

# Outer rectangle: AI
ai = plt.Rectangle(
    (0.5, 0.5), 9, 6,
    facecolor="lightblue",
    edgecolor="blue",
    linewidth=2
)
ax.add_patch(ai)

# Middle rectangle: ML
ml = plt.Rectangle(
    (1.5, 1.0), 7, 4.5,
    facecolor="lightgreen",
    edgecolor="green",
    linewidth=2
)
ax.add_patch(ml)

# Inner rectangle: Deep Learning
dl = plt.Rectangle(
    (2.5, 1.5), 5, 2.5,
    facecolor="moccasin",
    edgecolor="darkorange",
    linewidth=2
)
ax.add_patch(dl)

ax.text(0.8, 5.8, "Artificial Intelligence (AI)", fontsize=12)
ax.text(1.8, 4.8, "Machine Learning (ML)", fontsize=12)
ax.text(3.0, 2.5, "Deep Learning (DL)", fontsize=12)

ax.set_xlim(0, 10)
ax.set_ylim(0, 7)
ax.set_title("Relationship Between AI, ML and Deep Learning")
ax.axis("off")

plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier

# Sample data: [hours studied]
X = [[1], [2], [3], [4], [5], [6]]

# 0 = Fail, 1 = Pass
y = [0, 0, 0, 1, 1, 1]

# Create and train the model
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

# Predict for a student who studied 4 hours
hours = [[4]]
prediction = model.predict(hours)[0]

result = "Pass" if prediction == 1 else "Fail"

print("Hours studied:", hours[0][0])
print("Predicted result:", result)
```
